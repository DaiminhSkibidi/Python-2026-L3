```python
# Ex3 

num = int(input("Enter number? "))

def is_prime(num):
    # nums <= 1 are not prime
    if num <= 1:
        return False
        
    # Check all nums from 2 up to the square root of num
    for i in range(2, int(num**0.5) + 1): # int cuz range doesnt allow float
        if num % i == 0:
            return False  # if divisible -> not prime
            
    return True  # No factors found, it is prime

if is_prime(num) == True:
    print(f"{num} is a prime number")
else:
    print(f"{num} is NOT a prime number")
```

```python
# Ex4

num = int(input("Enter a number? "))
total = 0

for i in range(1, int(num**0.5) + 1):
    if num % i == 0: 
        total += i
        if i != 1:
            total += num // i # / gives float, // gives int, this add the pair.
            

if num == total and (num > 1):
    print(f"{num} is a perfect number")
else:
    print(f"{num} is a NOT perfect number")
```

```python
# Ex7

def remove_dollar_sign(s):
    result = ""  # new string
    for i in s:
        if i != '$':  
            result += i  # add each char that != $ into new string
    return result

s = "s3$$6$7"
print(remove_dollar_sign(s))  
```

```python
# Ex8
def extract_even(l):
    result = [] # new list
    for i in l:
        if i % 2 == 0: # append even num into new str
            result.append(i) 
    return result

l = [1, 4, 5, -1, 10]
print(extract_even(l)) 
```

```python
# Ex9
def factorial(n):
    # Base Case: 
    if n == 0 or n == 1:
        return 1
        
    # Recursive Case: 
    return n * factorial(n - 1)

print(factorial(5))  

```

```python
# Ex10
def get_divisors(n):
    divisors = []
    # Test every number from 1 to n
    for i in range(1, n + 1):
        if n % i == 0:
            divisors.append(i) # Add to list
    return divisors

print(get_divisors(12))  
```

```python
# Ex11
def distance(x1, y1, x2, y2):
    dx_squared = (x2 - x1) ** 2
    dy_squared = (y2 - y1) ** 2
    
    distance = (dx_squared + dy_squared) ** 0.5
    return distance

print(distance(1, 2, 4, 6)) 
```

```python
# Ex12

def print_pattern(m, n):
    for i in range(m):  # Loop through each row
        for j in range(n):# Loop through each col in that row
            # Check if current position is on any boundary line
            if i == 0 or i == m - 1 or j == 0 or j == n - 1:
                print("*", end=" ")
            else:
                print(" ", end=" ") # Empty space inside the box
        print() # Move to the next line

# m = 4, n = 5
print_pattern(4, 5)
```
